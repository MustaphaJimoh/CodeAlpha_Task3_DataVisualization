```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

print("Dataset loaded successfully!")
df.head()
```

```python
sns.set_style("whitegrid")

plt.figure()
sns.countplot(data=df, x='Survived', palette=['#e74c3c', '#2ecc71'])
plt.title('Titanic Survival Count', fontsize=16, fontweight='bold')
plt.xticks([0, 1], ['Did Not Survive', 'Survived'])
plt.xlabel('Survival Status')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
# Chart 2: Survival by Gender
plt.figure()
sns.countplot(data=df, x='Sex', hue='Survived', palette=['#e74c3c', '#2ecc71'])
plt.title('Survival by Gender', fontsize=16, fontweight='bold')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.legend(['Did Not Survive', 'Survived'])
plt.show()
```

```python
# Chart 3: Survival by Passenger Class
plt.figure()
sns.countplot(data=df, x='Pclass', hue='Survived', palette=['#e74c3c', '#2ecc71'])
plt.title('Survival by Passenger Class', fontsize=16, fontweight='bold')
plt.xlabel('Passenger Class (1=First, 2=Second, 3=Third)')
plt.ylabel('Number of Passengers')
plt.legend(['Did Not Survive', 'Survived'])
plt.show()
```

```python
# Chart 4: Age Distribution by Survival
plt.figure()
sns.histplot(data=df, x='Age', hue='Survived', bins=30,
             palette=['#e74c3c', '#2ecc71'], alpha=0.7)
plt.title('Age Distribution by Survival', fontsize=16, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Count')
plt.legend(['Did Not Survive', 'Survived'])
plt.show()
```

```python
# Chart 5: Fare Distribution by Survival
plt.figure()
sns.boxplot(data=df, x='Survived', y='Fare',
            palette=['#e74c3c', '#2ecc71'])
plt.title('Fare Distribution by Survival', fontsize=16, fontweight='bold')
plt.xticks([0, 1], ['Did Not Survive', 'Survived'])
plt.xlabel('Survival Status')
plt.ylabel('Fare Paid')
plt
```

```python
# Chart 6: Correlation Heatmap
plt.figure()
corr = df[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap', fontsize=16, fontweight='bold')
plt.show()
```
